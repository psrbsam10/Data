### Linier Algebra (Calvin - Samuel)

```python
import numpy as np

# Definisi vektor
x = np.array([1, 2])
y = np.array([3, 4])
z = np.array([5, 6])
zero = np.array([0, 0])

print("x =", x)
print("y =", y)
print("z =", z)
print("zero =", zero)
print()

# ------------------------------
# 1. Komutatif
# ------------------------------
print("1. Komutatif:")
print(f"x + y = ({x[0]}+{y[0]}, {x[1]}+{y[1]}) = ({x[0]+y[0]}, {x[1]+y[1]})")
print(f"y + x = ({y[0]}+{x[0]}, {y[1]}+{x[1]}) = ({y[0]+x[0]}, {y[1]+x[1]})")
print()

# ------------------------------
# 2. Asosiatif
# ------------------------------
print("2. Asosiatif:")
lhs = (x + y) + z
rhs = x + (y + z)
print(f"(x + y) + z = ({x[0]}+{y[0]}, {x[1]}+{y[1]}) + ({z[0]}, {z[1]}) "
      f"= ({x[0]+y[0]}+{z[0]}, {x[1]+y[1]}+{z[1]}) = {tuple(lhs)}")

print(f"x + (y + z) = ({x[0]}, {x[1]}) + ({y[0]}+{z[0]}, {y[1]}+{z[1]}) "
      f"= ({x[0]}+{y[0]+z[0]}, {x[1]}+{y[1]+z[1]}) = {tuple(rhs)}")
print()

# ------------------------------
# 3. Identitas (Zero Vector)
# ------------------------------
print("3. Identitas:")
lhs1 = x + zero
lhs2 = zero + x
print(f"x + 0 = ({x[0]}+0, {x[1]}+0) = ({lhs1[0]}, {lhs1[1]})")
print(f"0 + x = (0+{x[0]}, 0+{x[1]}) = ({lhs2[0]}, {lhs2[1]})")
print()

# ------------------------------
# 4. Invers
# ------------------------------
print("4. Invers:")
lhs = x - x
print(f"x - x = ({x[0]}-{x[0]}, {x[1]}-{x[1]}) = ({lhs[0]}, {lhs[1]})")
```

```python
import numpy as np

# Definisi vektor
x = np.array([2, 3])
y = np.array([4, 1])
z = np.array([1, -2])
zero = np.array([0, 0])

print("x =", x)
print("y =", y)
print("z =", z)
print("zero =", zero)

# ------------------------------
# 1. Komutatif: x + y = y + x
# ------------------------------
lhs = x + y   # sisi kiri
rhs = y + x   # sisi kanan
print("\n1. Komutatif:")
print("x + y =", lhs)
print("y + x =", rhs)
print("Hasil sama? ->", np.array_equal(lhs, rhs))

# ------------------------------
# 2. Asosiatif: (x + y) + z = x + (y + z)
# ------------------------------
lhs = (x + y) + z
rhs = x + (y + z)
print("\n2. Asosiatif:")
print("(x + y) + z =", lhs)
print("x + (y + z) =", rhs)
print("Hasil sama? ->", np.array_equal(lhs, rhs))

# ------------------------------
# 3. Identitas: x + 0 = 0 + x = x
# ------------------------------
lhs1 = x + zero
lhs2 = zero + x
print("\n3. Identitas:")
print("x + 0 =", lhs1)
print("0 + x =", lhs2)
print("Sama dengan x =", x, "? ->",
      np.array_equal(lhs1, x) and np.array_equal(lhs2, x))

# ------------------------------
# 4. Invers: x - x = 0
# ------------------------------
lhs = x - x
print("\n4. Invers:")
print("x - x =", lhs)
print("Sama dengan vektor nol =", zero, "? ->", np.array_equal(lhs, zero))
```
